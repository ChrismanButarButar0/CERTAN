```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

%matplotlib inline
```

```python
#2. Buatlah grafik 2 dimensi dari persamaan berikut
# x6+3x4y2-3x4+3x2y4-x2y3-6x2y2+3x2+y6 -3y4 +3y2 = 1

# Membuat grid X dan Y (rentang -1.5 sampai 1.5 agar kurva terlihat)
x = np.linspace(-1.5, 1.5, 500)
y = np.linspace(-1.5, 1.5, 500)
X, Y = np.meshgrid(x, y)

# Menghitung nilai Z dari persamaan
Z = (X**6 + 3*X**4*Y**2 - 3*X**4 + 3*X**2*Y**4 - X**2*Y**3 - 6*X**2*Y**2 + 
     3*X**2 + Y**6 - 3*Y**4 + 3*Y**2)

# Membuat plot kontur 2D pada level Z = 1
plt.figure(figsize=(8, 8))
contour = plt.contour(X, Y, Z, levels=[1], colors='blue', linewidths=3)
plt.title('Grafik 2D Persamaan Implisit (Tugas 2)', fontsize=14)
plt.xlabel('X')
plt.ylabel('Y')
plt.axis('equal') # Agar proporsi sumbu X dan Y seimbang
plt.grid(True, linestyle='--', alpha=0.6)

# Menandai kurva
plt.clabel(contour, inline=True, fontsize=10)
plt.show()
```
