```python
#1. Buatlah sebuah Matriks berukuran 5x5 dengan elemen random. Kemudian dari elemen yang dihasilkan perlihatkan nilai maksimum, nilai minimum dan
#elemen matriks yang di konstruksi.
matriks = np.random.randint(0, 100, size=(5, 5))

print("=== MATRIKS 5x5 ===")
print(matriks)

# Mencari nilai maksimum dan minimum
nilai_max = np.max(matriks)
nilai_min = np.min(matriks)

print(f"\nNilai Maksimum: {nilai_max}")
print(f"Nilai Minimum: {nilai_min}")

print("\nElemen-elemen matriks:")
print(matriks.flatten())
```
