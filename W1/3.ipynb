```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

%matplotlib inline
```

```python
# 3. Buatlah grafik 3 dimensi dari persamaan berikut
# (5y/4 - |x|)^2 = 1 - x^2

# Domain x harus antara -1 dan 1 agar akar kuadrat valid
x_vals = np.linspace(-1, 1, 100)
z_vals = np.linspace(-3, 3, 50) # Rentang Z untuk ekstrusi

# Membuat meshgrid 3D (X dan Z)
X_grid, Z_grid = np.meshgrid(x_vals, z_vals)

# Menghitung Y untuk dua cabang (plus dan minus)
Y_plus = (4/5) * (np.abs(X_grid) + np.sqrt(1 - X_grid**2))
Y_minus = (4/5) * (np.abs(X_grid) - np.sqrt(1 - X_grid**2))

# Plot 3D
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot permukaan cabang atas (plus) dengan transparansi
ax.plot_surface(X_grid, Y_plus, Z_grid, alpha=0.6, color='cyan', rstride=2, cstride=2)
# Plot permukaan cabang bawah (minus)
ax.plot_surface(X_grid, Y_minus, Z_grid, alpha=0.6, color='magenta', rstride=2, cstride=2)

ax.set_title('Grafik 3D Persamaan (Tugas 3)', fontsize=14)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.view_init(elev=20, azim=45) # Mengatur sudut pandang kamera 

plt.show()
```
