```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

%matplotlib inline
```

```python
#1. Buatlah sebuah Matriks berukuran 5x5 dengan elemen random. Kemudian dari elemen yang dihasilkan perlihatkan nilai maksimum, nilai minimum dan
#elemen matriks yang di konstruksi.
matriks = np.random.randint(0, 100, size=(5, 5))

print("=== MATRIKS 5x5 ===")
print(matriks)

# Mencari nilai maksimum dan minimum
nilai_max = np.max(matriks)
nilai_min = np.min(matriks)

print(f"\nNilai Maksimum: {nilai_max}")
print(f"Nilai Minimum: {nilai_min}")

print("\nElemen-elemen matriks:")
print(matriks.flatten())
```

```python
#2. Buatlah grafik 2 dimensi dari persamaan berikut
# x6+3x4y2-3x4+3x2y4-x2y3-6x2y2+3x2+y6 -3y4 +3y2 = 1

# Membuat grid X dan Y (rentang -1.5 sampai 1.5 agar kurva terlihat)
x = np.linspace(-1.5, 1.5, 500)
y = np.linspace(-1.5, 1.5, 500)
X, Y = np.meshgrid(x, y)

# Menghitung nilai Z dari persamaan
Z = (X**6 + 3*X**4*Y**2 - 3*X**4 + 3*X**2*Y**4 - X**2*Y**3 - 6*X**2*Y**2 + 
     3*X**2 + Y**6 - 3*Y**4 + 3*Y**2)

# Membuat plot kontur 2D pada level Z = 1
plt.figure(figsize=(8, 8))
contour = plt.contour(X, Y, Z, levels=[1], colors='blue', linewidths=3)
plt.title('Grafik 2D Persamaan Implisit (Tugas 2)', fontsize=14)
plt.xlabel('X')
plt.ylabel('Y')
plt.axis('equal') # Agar proporsi sumbu X dan Y seimbang
plt.grid(True, linestyle='--', alpha=0.6)

# Menandai kurva
plt.clabel(contour, inline=True, fontsize=10)
plt.show()
```

```python
# ============================================
# 3. GRAFIK 3 DIMENSI
# ============================================

import numpy as np
import matplotlib.pyplot as plt

# Membuat range x dan y
x = np.linspace(-4, 4, 200)
y = np.linspace(-4, 4, 200)

# Membuat meshgrid
X, Y = np.meshgrid(x, y)

# Persamaan:
# z = (5y/4 - |x|)^2 + x^2

Z = ((5 * Y / 4) - np.abs(X))**2 + X**2

# Membuat grafik 3D
fig = plt.figure(figsize=(10, 8))

ax = fig.add_subplot(111, projection="3d")

# Plot permukaan
permukaan = ax.plot_surface(
    X,
    Y,
    Z,
    cmap="viridis",
    edgecolor="none"
)

# Label sumbu
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

# Judul
ax.set_title(r"$z = (5y/4 - |x|)^2 + x^2$")

# Colorbar
fig.colorbar(
    permukaan,
    ax=ax,
    shrink=0.6,
    label="Nilai Z"
)

plt.show()
```
